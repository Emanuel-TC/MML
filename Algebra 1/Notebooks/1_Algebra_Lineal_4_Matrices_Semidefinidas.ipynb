## Definición

#### Matriz definida positiva

Una matriz $\mathbf{A} \in \mathbb{R}^{n \times n}$ es definida positiva, que se denota por $\mathbf{A} \succ \boldsymbol{0}$, si

\begin{equation}
{\mathbf x}^T {\mathbf A} {\mathbf x} > 0, \quad \forall {\mathbf x} \neq {\boldsymbol 0} 
\end{equation}

#### Matriz semidefinida positiva

Una matriz ${\mathbf A} \in \mathbb{R}^{n \times n}$ es semidefinida positiva, que se denota por ${\mathbf A} \succeq {\boldsymbol 0}$, si

\begin{equation}
{\mathbf x}^T {\mathbf A} {\mathbf x} \geq 0
\end{equation}

#### Matriz definida negativa

Una matriz ${\mathbf A} \in \mathbb{R}^{n \times n}$ es definida negativa, que se denota por ${\mathbf A} \prec {\boldsymbol 0}$, si

\begin{equation}
{\mathbf x}^T {\mathbf A} {\mathbf x} < 0, \quad \forall {\mathbf x} \neq {\boldsymbol 0} 
\end{equation}

#### Matriz semidefinida negativa

Una matriz ${\mathbf A} \in \mathbb{R}^{n \times n}$ es semidefinida negativa, que se denota por ${\mathbf A} \preceq {\boldsymbol 0}$, si

\begin{equation}
{\mathbf x}^T {\mathbf A} {\mathbf x} \leq 0
\end{equation}

##### Ordenamiento de matrices

Esta definición nos permite "ordenar" matrices

\begin{equation}
{\mathbf A} \succeq {\mathbf B} \Leftrightarrow  {\mathbf A} - {\mathbf B} \succeq {\boldsymbol 0}
\end{equation}

## Propiedades

1. Si ${\mathbf A} \succ {\boldsymbol 0}$, entonces $\lambda_i({\mathbf A}) > 0$ (esta propiedad se puede extrapolar a ${\mathbf A} \succeq {\boldsymbol 0}, {\mathbf A} \prec {\boldsymbol 0}, {\mathbf A} \preceq {\boldsymbol 0}$)
2. Si ${\mathbf A} \succ {\boldsymbol 0}$, entonces ${\mathbf A}^{-1} \succ {\boldsymbol 0}$
3. Para cualquier matriz ${\mathbf B}$, ${\mathbf A} = {\mathbf B} {\mathbf B}^T \succeq {\boldsymbol 0}$

Como consecuencia de la Propiedad 1, tenemos

- Si ${\mathbf A} \succ {\boldsymbol 0}$, tenemos $\operatorname{tr}({\mathbf A}) > 0$
- Si ${\mathbf A} \succ {\boldsymbol 0}$, tenemos $\operatorname{det}({\mathbf A}) > 0$

```python
# Matriz y vector de ejemplo
B = np.array([[1,3,2],[5,4,6],[7,8,9]])
A = B @ B.T

# Autovalores de A
Lambda, _ = np.linalg.eig(A)
print(Lambda)

# Traza y determinante de A
print(np.trace(A))
print(np.linalg.det(A))
```

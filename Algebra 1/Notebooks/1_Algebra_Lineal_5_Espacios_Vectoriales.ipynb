## Espacios y subespacios

#### ¿Qué es un espacio vectorial?

Es un tipo de estructura algebraica donde los elementos son vectores y tiene que cumplir una serie de axiomas, que principalmente tienen que ver con la suma de vectores y multiplicaciones por escalares.

Durante el curso ya han aparecido espacios vectoriales. Concretamente, hemos considerado el espacio $\mathbb{R}^{n}$, el cual está formado por todos los vectores de tamaño $n$, y es el que consideraremos a continuación.

Para continuar, necesitamos una serie de definiciones.

#### Combinación lineal de vectores y alcance

Dados $p$ vectores de $\mathbb{R}^{n}$, $\mathbf{a}_1, \ldots, \mathbf{a}_p \in \mathbb{R}^{n}$, y $p$ escalares, $b_1, \ldots, b_p \in \mathbb{R}$, la combinación lineal de estos se define como

\begin{equation}
{\mathbf c} = b_1 \left[\begin{array}{c}
\\
\\
{\mathbf a}_1 \\
\\
\\
\end{array}
\right]
+ \cdots +
b_p \left[\begin{array}{c}
\\
\\
{\mathbf a}_p \\
\\
\\
\end{array}
\right] = \sum_{l = 1}^{p} {\mathbf a}_l b_l
\end{equation}

donde ${\mathbf c} \in \mathbb{R}^{n}$. Es decir, una combinación lineal de vectores pertenece al mismo espacio que los vectores. Si definimos la matriz

\begin{equation}
{\mathbf A} = 
\left[
\begin{array}{c|c|c|c}
 &  &  &  \\
&  &  &  \\
{\mathbf a}_1 & {\mathbf a}_2 & \cdots & {\mathbf a}_p \\
&  &  &  \\
&  &  &  
\end{array}
\right]
\end{equation}

podemos escribir

\begin{equation}
\left[
\begin{array}{c}
\\
\\
{\mathbf c} \\
\\
\\
\end{array}
\right] = 
{\mathbf A}
\begin{bmatrix}
b_{1} \\
b_{2} \\
\vdots \\
b_{p}
\end{bmatrix} =
{\mathbf A} {\mathbf b}
\end{equation}

Retomemos la figura de la sección de "Repaso de cálculo matricial" y consireremos el espacio vectorial $\langle {\mathbf A} \rangle$, que es el plano gris de la figura y que está formado por todos los vectores que viven en dicho plano. Es decir, todos los vectores ${\mathbf c} = {\mathbf A} {\mathbf b}$ para todos los posibles ${\mathbf b} \in \mathbb{R}^p$

<img 
    style="display: block; 
           margin-left: auto;
           margin-right: auto;
           width: 40%;"
    src='./figures/span.svg'/>
</img>

Cualquier vector de $\langle {\mathbf A} \rangle$ se puede escribir como la combinación lineal de los vectores ${\mathbf a}_1$ y ${\mathbf a}_2$. Es decir, este espacio vectorial ha sido generado por la matriz ${\mathbf A}$. Este espacio vectorial generado por las columnas de ${\mathbf A}$ se denomina **alcance** de ${\mathbf A}$, $\operatorname{range}({\mathbf A})$ o $\operatorname{span}({\mathbf A})$

#### Independencia lineal y espacio nulo de una matriz

Un subconjunto de vectores de un espacio vectorial son linealmente independientes si ninguno se puede escribir como una combinación lineal del resto. Matemáticamente, un conjunto de $p$ vectores sería linealmente dependiente si

\begin{equation}
{\boldsymbol 0} = b_1 \left[\begin{array}{c}
\\
\\
{\mathbf a}_1 \\
\\
\\
\end{array}
\right]
+ \cdots +
b_p \left[\begin{array}{c}
\\
\\
{\mathbf a}_p \\
\\
\\
\end{array}
\right]
\end{equation}

Por ejemplo, cualesquiera dos vectores de $\langle {\mathbf A} \rangle$ son linealmente independientes (salvo que uno sea una versión escalada del otro). No obstante, si elegimos tres vectores de $\langle {\mathbf A} \rangle$ no serán linealmente independientes.

El **espacio nulo** de ${\mathbf A}$, $\operatorname{null}({\mathbf A})$, se define como el espacio vectorial generado por los vectores ${\mathbf b}$ tales que
\begin{equation}
{\boldsymbol 0} = b_1 \left[\begin{array}{c}
\\
\\
{\mathbf a}_1 \\
\\
\\
\end{array}
\right]
+ \cdots +
b_p \left[\begin{array}{c}
\\
\\
{\mathbf a}_p \\
\\
\\
\end{array}
\right] = {\mathbf A} {\mathbf b}
\end{equation}

#### Subespacio vectorial

Es un subconjunto de un espacio vectorial, tal que la suma de dos elementos del subespacio o la multiplicación de uno por un escalar pertenecen al subespacio. Por ejemplo, el espacio vectorial $\langle {\mathbf A} \rangle$ de la figura es un subespacio de $\mathbb{R}^3$.

#### Base del espacio vectorial

Una base de un subespacio vectorial es el conjunto de vectores linealmente independientes que generan dicho subespacio. En el ejemplo anterior, ${\mathbf a}_1$ y ${\mathbf a}_2$ serían una base de $\langle {\mathbf A} \rangle$. No obstante, ${\mathbf a}_1$ y $c {\mathbf a}_1$ no lo serían por no ser linealmente independientes. Tampoco sería una base si añadimos un tercer vector a ${\mathbf a}_1$ y ${\mathbf a}_2$

##### Base ortonormal

Es habitual que se trabaje con bases ortogonales. Es decir, bases cuyos elementos son ortogonales por parejas: ${\mathbf a}_i^T {\mathbf a}_j = 0, i \neq j$. De esta manera se asegura la independencia lineal. Sin embargo, es todavía más común trabajar con bases ortonormales donde los elementos son ortogonales y, además, tienen norma unidad: ${\mathbf a}_i^T {\mathbf a}_j = 0, i \neq j$ y $\| {\mathbf a}_i \|_2 = 1$. Por lo tanto, la matriz compuesta por los elementos de la base

\begin{equation}
{\mathbf A} = 
\left[
\begin{array}{c|c|c|c}
 &  &  &  \\
&  &  &  \\
{\mathbf a}_1 & {\mathbf a}_2 & \cdots & {\mathbf a}_p \\
&  &  &  \\
&  &  &  
\end{array}
\right]
\end{equation}

cumple

\begin{equation}
{\mathbf A}^T {\mathbf A} = {\mathbf I}
\end{equation}

Es decir, ${\mathbf A}$ es un *slice* de una matriz ortogonal. Finalmente, si consideramos el espacio vectorial $\mathbb{R}^n$, en lugar de un subespacio, la matriz ${\mathbf A}$ será una ortogonal

#### Dimensión del espacio vectorial y rango de una matriz

La dimensión del espacio (o subespacio) vectorial es el número de vectores que forman la base de dicho espacio (o subespacio). El rango de una matriz es la dimensión del espacio vectorial generado por dicha matriz. Es importante destacar que no tiene porque coincidir con el número de columnas ya que estas pueden ser linealmente dependientes.

##### Condiciones equivalentes

Para una matriz ${\mathbf A} \in \mathbb{R}^{n \times n}$, las siguientes condiciones son equivalentes

- ${\mathbf A}$ es regular/no es singular/tiene inversa ${\mathbf A}^{-1}$
- $\operatorname{rank}({\mathbf A}) = n$
- $\operatorname{range}({\mathbf A}) = \mathbb{R}^{n}$
- $\operatorname{null}({\mathbf A}) = \{\emptyset\}$
- $\operatorname{det}({\mathbf A}) \neq 0$
- $0$ no es un autovalor de ${\mathbf A}$ ni un valor singular

```python
# Matriz de ejemplo
# Las columnas de A son una base para un subespacio de dimension 2 del espacio tri-dimensional
A = np.array([[1,3],[5,4],[7,8]])

# Base ortonormal para el subespacio generado por A
U = sp.linalg.orth(A)
print(U.T @ U)

# Base ortonormal del espacio nulo de A
U = sp.linalg.null_space(A)
print(U)

print(np.linalg.matrix_rank(A))

# Base ortonormal del espacio nulo de A^T
U = sp.linalg.null_space(A.T)
print(U)

print(np.linalg.matrix_rank(A.T))
```

##### ¿Qué ocurre si $\| {\mathbf a} \| \neq 1$?

En este caso, lo que nos interesa es la dirección marcada por ${\mathbf a}$. Por lo tanto, podemos normalizarlo para obtener la dirección

\begin{equation}
    \tilde{{\mathbf a}} = \frac{{\mathbf a}}{\| {\mathbf a} \|} = \frac{{\mathbf a}}{\sqrt{{\mathbf a}^T {\mathbf a}}}
\end{equation}

Entonces, el producto escalar $\tilde{{\mathbf a}}^T {\mathbf b}$ mide la longitud de la proyección de ${\mathbf b}$ sobre la dirección indicada por ${\mathbf a}$ y la proyección sería

\begin{equation}
    {\mathbf c} = (\tilde{{\mathbf a}}^T {\mathbf b}) \tilde{{\mathbf a}} = \frac{{\mathbf a}^T {\mathbf b}}{{\mathbf a}^T {\mathbf a}} {\mathbf a} = {\mathbf a} ({\mathbf a}^T {\mathbf a})^{-1} {\mathbf a}^T {\mathbf b}
\end{equation}

#### Proyección sobre un subespacio arbitrario

Consideremos, en primer lugar, que tenemos una base ortonormal para el subespacio dada por ${\mathbf U} = [{\mathbf u}_1 \ \cdots \ {\mathbf u}_p]$, como muestra la siguiente figura en el caso de $p = 2$

<img 
    style="display: block; 
           margin-left: auto;
           margin-right: auto;
           width: 40%;"
    src='./figures/proyeccion.svg'/>
</img>

Al ser una base ortonormal, sabemos que $\|{\mathbf u}_i\| = 1$ y ${\mathbf u}_i^T {\mathbf u}_j = 0, i \neq j$. Como $\|{\mathbf u}_i\| = 1$, el producto escalar ${\mathbf u}_i^T {\mathbf b}$ mide la longitud de la proyección de ${\mathbf b}$ sobre ${\mathbf u}_i$. Al igual que en el caso unidimensional, si lo que queremos es la proyección, no sólo su longitud, es necesario multiplicar por ${\mathbf u}_i$. Es decir, ${\mathbf u}_i {\mathbf u}_i^T {\mathbf b}$ será la proyección de ${\mathbf b}$ sobre ${\mathbf u}_i$. Finalmente, la suma de todas estas proyecciones será la proyección sobre el plano $\langle {\mathbf U} \rangle$, dada por

\begin{equation}
{\mathbf u}_1^T {\mathbf b} \left[\begin{array}{c}
\\
\\
{\mathbf u}_1 \\
\\
\\
\end{array}
\right]
+ \cdots +
{\mathbf u}_p^T {\mathbf b} \left[\begin{array}{c}
\\
\\
{\mathbf u}_p \\
\\
\\
\end{array}
\right] = 
\left[
\begin{array}{c|c|c|c}
 &  &  &  \\
&  &  &  \\
{\mathbf u}_1 & {\mathbf u}_2 & \cdots & {\mathbf u}_p \\
&  &  &  \\
&  &  &  
\end{array}
\right] \begin{bmatrix}
{\mathbf u}_1^T {\mathbf b} \\
{\mathbf u}_2^T {\mathbf b} \\
\vdots \\
{\mathbf u}_p^T {\mathbf b}
\end{bmatrix} = {\mathbf U} {\mathbf U}^T {\mathbf b} = {\mathbf P} {\mathbf b}
\end{equation}

donde ${\mathbf P} = {\mathbf U} {\mathbf U}^T$ es la matriz de proyección. Nótese que si ${\mathbf U}$ es cuadrada, es ortogonal, y ${\mathbf P} = {\mathbf U} {\mathbf U}^T = {\mathbf I}$. Por lo que, ${\mathbf P} {\mathbf b} = {\mathbf b}$. Es decir, sólo se obtiene algo diferente si se proyecta sobre un subespacio de dimensión inferior. O lo que es lo mismo, proyectar sobre el espacio vectorial original devuelve el mismo vector

##### ¿Qué ocurre si ${\mathbf A}^T {\mathbf A} \neq {\mathbf I}$?

En este caso, podemos ortonormalizar ${\mathbf A}$

\begin{equation}
    {\mathbf U} = {\mathbf A} ({\mathbf A}^T {\mathbf A})^{-1/2}
\end{equation}

donde $({\mathbf A}^T {\mathbf A})^{-1/2}$ es la inversa de la matriz raíz cuadrada. En este caso, la matriz de proyección será

\begin{equation}
    {\mathbf P} = {\mathbf U} {\mathbf U}^T = {\mathbf A} ({\mathbf A}^T {\mathbf A})^{-1/2} ({\mathbf A}^T {\mathbf A})^{-1/2} {\mathbf A}^T = {\mathbf A} ({\mathbf A}^T {\mathbf A})^{-1} {\mathbf A}^T
\end{equation}

donde se ha tenido en cuenta que ${\mathbf A}^T {\mathbf A}$ es simétrica

#### Proyector complementario

El proyector complementario al subespacio $\langle {\mathbf U} \rangle$ proyecta sobre el espacio nulo de ${\mathbf U} {\mathbf U}^T$. Es decir, es el proyector sobre el subespacio perpendicular, denotado como $\langle {\mathbf U}^{\perp} \rangle$. Esto se puede ver en la siguiente figura.

<img 
    style="display: block; 
           margin-left: auto;
           margin-right: auto;
           width: 40%;"
    src='./figures/proyeccion_perp.svg'/>
</img>

Concretamente el proyector complementario se calcula como

\begin{equation}
    {\mathbf P}^{\perp} = {\mathbf U}^{\perp} ({\mathbf U}^{\perp})^T = {\mathbf I} - {\mathbf U} {\mathbf U}^T = {\mathbf I} - {\mathbf P}
\end{equation}

Por lo tanto, un proyector y su complementario separan el espacio en dos subespacios

#### Propiedades de los proyectores

- Las matrices de proyección son simétricas: ${\mathbf P} = {\mathbf P}^T$ (esto ocurre porque consideramos proyectores ortogonales)
- Las matrices de proyección son idempotentes: ${\mathbf P} = {\mathbf P}^2$
- Las matrices de proyección tienen $p$ autovalores iguales a $1$ y $n - p$ autovalores iguales a $0$
- ${\mathbf P}^{\perp} {\mathbf P} = {\mathbf P} {\mathbf P}^{\perp} = {\boldsymbol 0}$
